```python
import os, shutil, subprocess
from pathlib import Path
repo = Path('/kaggle/working/zero-to-empire')
if repo.exists(): shutil.rmtree(repo)
subprocess.run(['git','clone','--depth','1','https://github.com/dbrckk/zero-to-empire.git',str(repo)], check=True)
os.chdir(repo)
print('repo', subprocess.check_output(['git','rev-parse','--short','HEAD'], text=True).strip())
```

```python
COUNT = 60
subprocess.run(['python','tools/sprites/kaggle_sprite_factory.py','--kind','ALL','--count',str(COUNT)], check=True)
```

```python
subprocess.run(['python','tools/sprites/build_sprite_contact_sheet.py','--input','art/incoming/final-sprites','--output','/kaggle/working/kaggle-sprite-contact-sheet.png','--report','/kaggle/working/kaggle-sprite-qa.json'], check=True)
from IPython.display import display
from PIL import Image
display(Image.open('/kaggle/working/kaggle-sprite-contact-sheet.png'))
```

```python
out = shutil.make_archive('/kaggle/working/zero-to-empire-kaggle-candidates','zip',root_dir='art/incoming/final-sprites')
print('OUTPUT_ZIP=', out)
```
